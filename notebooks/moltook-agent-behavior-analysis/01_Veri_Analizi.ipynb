```python
import pandas as pd
import ast
import matplotlib.pyplot as plt
import seaborn as sns
import os

# 1. DOSYA YOLU KONTROLÜ VE YÜKLEME
file_path = 'moltbook_temiz_final.csv'

if not os.path.exists(file_path):
    print(f"❌ HATA: {file_path} bulunamadı!")
else:
    print("1. Temizlenmiş veri seti yükleniyor...")
    try:
        df = pd.read_csv(file_path, low_memory=False)
    except Exception as e:
        print(f"❌ Dosya okuma hatası: {e}")

    # Konu harflerini isimlendiriyoruz
    topic_mapping = {
        'A': 'A (Genel Tartışma)', 'B': 'B (Politika)', 'C': 'C (Ekonomi)',
        'D': 'D (Teknoloji)', 'E': 'E (Spor)', 'F': 'F (Kültür & Sanat)',
        'G': 'G (Sağlık & Yaşam)', 'H': 'H (Eğitim & Bilim)', 'I': 'I (Diğer)'
    }
    df['topic_display'] = df['topic_label'].map(topic_mapping).fillna(df['topic_label'])

    # 2. TARİH VERİLERİNİ AYIKLAMA
    df['created_at'] = pd.to_datetime(df['timestamp'], errors='coerce')
    df = df.dropna(subset=['created_at'])
    df['date'] = df['created_at'].dt.date

    sns.set_theme(style="whitegrid")
    print("\n3. Grafikler oluşturuluyor ve kaydediliyor...")

    # --- GRAFİK 1: Günlük Aktivite ---
    plt.figure(figsize=(12, 6))
    daily_counts = df.groupby('date').size().reset_index(name='post_count')
    sns.lineplot(data=daily_counts, x='date', y='post_count', marker='o', color='dodgerblue')
    plt.title('Günlük Gönderi Aktivitesi', fontsize=14)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig('plots/01_daily_activity.png', dpi=150) 
    plt.close()

    # --- GRAFİK 2: Konu Dağılımı (Güncellenmiş Etiketlerle) ---
    plt.figure(figsize=(12, 6))
    topic_counts = df['topic_display'].value_counts().reset_index()
    topic_counts.columns = ['topic_display', 'count']
    sns.barplot(data=topic_counts, x='topic_display', y='count', palette='viridis')
    plt.title('Konu Dağılımı', fontsize=14)
    plt.xlabel('Konu (Harf ve İsim)', fontsize=12)
    plt.ylabel('Gönderi Sayısı', fontsize=12)
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.savefig('plots/02_topic_distribution.png', dpi=150)
    plt.close()
    print("- 'konu_dagilimi.png' (Güncellendi) kaydedildi.")

    # --- GRAFİK 3: Toksisite Dağılımı ---
    plt.figure(figsize=(8, 5))
    toxic_counts = df['toxic_level'].value_counts().sort_index().reset_index()
    toxic_counts.columns = ['toxic_level', 'count']
    sns.barplot(data=toxic_counts, x='toxic_level', y='count', palette='Reds')
    plt.title('Toksisite Seviyesi Dağılımı', fontsize=14)
    plt.tight_layout()
    plt.savefig('plots/03_toxicity_distribution.png', dpi=150)
    plt.close()

    print("\n✅ İlk temel grafikler güncel etiketlerle oluşturuldu.")
```

```python
import pandas as pd
import ast
import matplotlib.pyplot as plt
import seaborn as sns
import os

def run_agent_analysis():
    # 1. DOSYA YOLU VE YÜKLEME (Mutlak yol kullanıyoruz)
    file_path = 'moltbook_temiz.csv'

    if not os.path.exists(file_path):
        print(f"❌ HATA: {file_path} bulunamadı! Lütfen yüklemenin bittiğinden emin ol.")
        return

    print("1. Veri seti analiz için yükleniyor...")
    # Düşük bellek uyarısını kapatmak ve hatalı satırları atlamak için parametreler ekledik
    df = pd.read_csv(file_path, low_memory=False, on_bad_lines='warn')

    # 2. VERİ AYIKLAMA (Tek geçişte 'post' sütununu işle)
    print("2. 'post' sütunundan tarih ve yazar bilgileri çekiliyor...")

    def parse_post_data(post_str):
        try:
            # String'i sözlüğe çevir
            p_dict = ast.literal_eval(post_str)
            # Hem tarihi hem de id'yi (author_id) döndür
            return p_dict.get('created_at'), p_dict.get('id')
        except:
            return None, None

    # Yeni sütunları oluştur (Zip kullanarak iki değeri aynı anda dağıtıyoruz)
    extracted = df['post'].apply(parse_post_data)
    df[['created_at_raw', 'author_id']] = pd.DataFrame(extracted.tolist(), index=df.index)

    # Tarih dönüşümü ve temizlik
    df['created_at'] = pd.to_datetime(df['created_at_raw'], errors='coerce')
    df = df.dropna(subset=['author_id']) # ID'si olmayan satırları temizle

    # 3. EN AKTİF 20 AJAN ANALİZİ
    print("3. En aktif ajanlar hesaplanıyor...")
    # ID'leri string'e çeviriyoruz (Grafikte sayısal karmaşa olmaması için)
    df['author_id'] = df['author_id'].astype(str)

    top20 = df['author_id'].value_counts().head(20).reset_index()
    top20.columns = ['author_id', 'post_count']

    # 4. GÖRSELLEŞTİRME
    plt.figure(figsize=(12, 8))
    sns.set_theme(style="whitegrid")

    # Barplot oluşturma
    sns.barplot(data=top20, x='post_count', y='author_id',
                hue='author_id', palette='Blues_r', legend=False)

    plt.title('En Aktif 20 Ajan (Gönderi Sayısına Göre)', fontsize=14, fontweight='bold')
    plt.xlabel('Toplam Gönderi Sayısı', fontsize=12)
    plt.ylabel('Ajan ID', fontsize=12)

    plt.tight_layout()
    plt.savefig('plots/15_top20_active_agents.png', dpi=150)
    plt.show()

    # 5. SONUÇLARI YAZDIR
    print("\n📊 --- En Aktif 5 Ajan Özeti ---")
    print(top20.head(5).to_string(index=False))
    print("\n✅ Analiz tamamlandı. 'en_aktif_20_ajan.png' kaydedildi.")

if __name__ == "__main__":
    run_agent_analysis()
```

Bu bölümdeki grafikler, teknik detaylardan ziyade verinin genel hikayesini daha şık ve anlaşılır bir dille anlatmak üzere tasarlanmıştır.

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

try:
    df = pd.read_csv('moltbook_temiz_final.csv', low_memory=False)
except Exception as e:
    print(f"❌ HATA: {e}")

topic_mapping = {
    'A': 'A (Genel Tartışma)', 'B': 'B (Politika)', 'C': 'C (Ekonomi)',
    'D': 'D (Teknoloji)', 'E': 'E (Spor)', 'F': 'F (Kültür & Sanat)',
    'G': 'G (Sağlık & Yaşam)', 'H': 'H (Eğitim & Bilim)', 'I': 'I (Diğer)'
}
df['topic_display'] = df['topic_label'].map(topic_mapping).fillna(df['topic_label'])
df['date'] = pd.to_datetime(df['timestamp'], errors='coerce').dt.date
df = df.dropna(subset=['date'])

sns.set_theme(style="white", palette="pastel")

# 1. Konu Yoğunluğu
plt.figure(figsize=(16, 8))
pivot_df = df.groupby(['date', 'topic_display']).size().unstack(fill_value=0)
pivot_df.plot(kind='area', stacked=True, alpha=0.7, figsize=(16, 8), colormap='tab10')
plt.title('Zaman İçinde Konu Dağılımı', fontsize=16, fontweight='bold')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.savefig('plots/10_topic_area_over_time.png', dpi=150)
plt.show()

print("\n--- 📊 GRAFİK RAPORU: KONU DAĞILIMI ---")
print("1. Platformun ana gövdesini 'Genel Tartışma' ve 'Politika' oluşturuyor.")
print("2. Belirli günlerde 'Teknoloji' ve 'Ekonomi' alanında hacim artışları gözlemleniyor.")
print("3. 'Kültür' ve 'Sağlık' gibi niş alanlar daha stabil ve düşük hacimli bir seyir izliyor.")

# 2. Toksisite Yoğunluğu
plt.figure(figsize=(14, 7))
sns.violinplot(data=df, x='topic_display', y='toxic_level', inner="quart", palette="muted")
plt.title('Konu Bazlı Toksisite Yoğunluğu', fontsize=16, fontweight='bold')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig('plots/08_toxicity_violin_analysis.png', dpi=150)
plt.show()

print("\n--- 📊 GRAFİK RAPORU: TOKSİSİTE ANALİZİ ---")
print("1. 'Politika' ve 'Eğitim & Bilim' konularında toksisite yelpazesi oldukça geniş (yüksek risk).")
print("2. 'Genel Tartışma' kategorisinde yoğunluk orta seviyede ancak hacim nedeniyle dikkat gerektiriyor.")
print("3. 'Kültür & Sanat' kategorisi platformun en 'nezaketli' ve güvenli alanı olarak saptanmıştır.")

# 3. Aktivite Trendi
plt.figure(figsize=(14, 6))
daily_counts = df.groupby('date').size().reset_index(name='count')
daily_counts['rolling_avg'] = daily_counts['count'].rolling(window=7).mean()
sns.lineplot(data=daily_counts, x='date', y='count', color='lightgray', alpha=0.5)
sns.lineplot(data=daily_counts, x='date', y='rolling_avg', color='darkblue', linewidth=2.5)
plt.title('Genel Aktivite Trendi', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.savefig('plots/11_activity_moving_average.png', dpi=150)
plt.show()

print("\n--- 📊 GRAFİK RAPORU: AKTİVİTE TRENDİ ---")
print("1. 7 günlük hareketli ortalama, platformun istikrarlı bir büyüme evresinde olduğunu gösteriyor.")
print("2. Günlük dalgalanmaların (gri çizgi) mavi ortalamaya yakınlığı, aktivitenin organik olduğunu kanıtlıyor.")
print("3. Son dönemdeki yukarı yönlü ivme, kullanıcı etkileşiminin arttığını işaret ediyor.")
```
